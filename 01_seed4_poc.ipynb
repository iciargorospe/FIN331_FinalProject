```python
# Import packages
import pandas as pd
import openassetpricing as oap
from importlib.metadata import version

# Specific version of financial data we are using (for reproducibility)
DATA_RELEASE = 202610

openap = oap.OpenAP(DATA_RELEASE)

# For documentation purposes
print("Open Asset Pricing release:", DATA_RELEASE)
print("Python package version:", version("openassetpricing"))
```

```python
# Download anomaly documentation
signaldoc = openap.dl_signal_doc("pandas")

# Inspect dataset
print('Dimensions:', signaldoc.shape)

print('Columns:', signaldoc.columns)

print(signaldoc.describe())

display(signaldoc.head())
```

```python
# Five proof-of-concept anomalies
signallist = ['BM', 'IndIPO', 'Mom12m', 'Accruals', 'AssetGrowth']

# Download portfolio returns
port_data = openap.dl_port('op', 'pandas', signallist)

# Inspect dataset
print('Dimensions:', port_data.shape)

print('Columns:', port_data.columns)

print('Portfolio types:', port_data['port'].value_counts())

print(port_data.describe())

display(port_data.head())
```

```python
# Look at documentation for poc anomalies
columns = ['Acronym', 'LongDescription', 'Year', 'Cat.Economic', 'Cat.Data', 'SampleStartYear', 'SampleEndYear']

display(signaldoc.loc[signaldoc['Acronym'].isin(signallist), columns])
```

```python
# Keep only long-short portfolio returns
ls_returns = port_data[port_data['port'] == 'LS']

# Inspect available observations for each anomaly
summary = ls_returns.groupby('signalname').agg(
    observations=('ret', 'count'),
    first_month=('date', 'min'),
    last_month=('date', 'max'),
    avg_monthly_ret=('ret', 'mean')
)

display(summary)

display(ls_returns.head())
```

```python
# Build proof-of-concept dataset

# Select relevant metadata columns
metadata = signaldoc.loc[signaldoc['Acronym'].isin(signallist), ['Acronym', 'Year', 'Cat.Economic', 'Cat.Data', 'SampleStartYear', 'SampleEndYear']]

# Rename columns
metadata = metadata.rename(columns={
    'Acronym': 'signalname',
    'Year': 'publication_year',
    'Cat.Economic': 'economic_category',
    'Cat.Data': 'data_category',
    'SampleStartYear': 'sample_start_year',
    'SampleEndYear': 'sample_end_year'
})

# Keep long-short returns
returns = ls_returns[['signalname', 'date', 'ret']].copy()

returns['date'] = pd.to_datetime(returns['date'])

# Merge returns with metadata
poc_data = pd.merge(returns,
    metadata,
    on='signalname',
    how='left',
    validate='m:1',
    indicator=True
)

# Verify that every return matched its metadata and drop merge column
assert (poc_data['_merge'] == 'both').all()
poc_data = poc_data.drop(columns='_merge')

# Create month and observation year columns
poc_data['month'] = (poc_data['date'].dt.to_period('M').astype(str))
poc_data['observation_year'] = poc_data["date"].dt.year

# Classify observations using the original research sample and publication year

poc_data['period'] = 'Before Original Sample'

# Original sample used in the academic paper
poc_data.loc[
    (poc_data['observation_year'] >= poc_data['sample_start_year']) &
    (poc_data['observation_year'] <= poc_data['sample_end_year']),
    'period'
] = 'Original Sample'

# Time between the end of the original sample and publication
poc_data.loc[
    (poc_data['observation_year'] > poc_data['sample_end_year']) &
    (poc_data['observation_year'] < poc_data['publication_year']),
    'period'
] = 'Pre-Publication Gap'

# Publication year
poc_data.loc[
    poc_data['observation_year'] == poc_data['publication_year'],
    'period'
] = 'Publication Year'

# After publication
poc_data.loc[
    poc_data['observation_year'] > poc_data['publication_year'],
    'period'
] = 'Post-Publication'

# Numeric indicator: 0 = before publication, 1 = after
# Excluded periods remain missing
poc_data['post_publication'] = pd.Series(
    pd.NA, index=poc_data.index, dtype='Int64'
)

poc_data.loc[
    poc_data['period'].isin(['Original Sample', 'Pre-Publication Gap']),
    'post_publication'
] = 0

poc_data.loc[
    poc_data['period'] == 'Post-Publication',
    'post_publication'
] = 1


#Organize final columns
final_columns = ['signalname', 'month', 'ret', 'publication_year', 'period', 'post_publication', 'economic_category', 'data_category', 'sample_start_year', 'sample_end_year']

poc_data = (poc_data[final_columns].sort_values(['signalname', 'month']).reset_index(drop=True))

# Validate data
# Verify one row per anomaly-month
assert not poc_data.duplicated(subset=['signalname', 'month']).any()

# Verify that every anomaly has a publication year
assert poc_data['publication_year'].notna().all()

print("Data validation checks passed!")

# Preview actual dataset
print('Final dataset dimensions:', poc_data.shape)
display(poc_data.head(10))

# Check number of observations in each period
display(poc_data.groupby(['signalname', 'period']).size())
```

```python
# Validate period classifications using Momentum as an example

months_to_check = [
    '1963-12',  # Before original sample
    '1964-01',  # Start of original sample
    '1989-12',  # End of original sample
    '1990-01',  # Start of pre-publication gap
    '1992-12',  # End of pre-publication gap
    '1993-01',  # Publication year
    '1994-01'   # Start of post-publication
]

validation_sample = poc_data.loc[
    (poc_data['signalname'] == 'Mom12m') &
    (poc_data['month'].isin(months_to_check)),
    ['month', 'ret', 'period', 'post_publication']
].sort_values('month')

display(validation_sample)

# Verify that all seven observations have the correct classification
expected_periods = [
    'Before Original Sample',
    'Original Sample',
    'Original Sample',
    'Pre-Publication Gap',
    'Pre-Publication Gap',
    'Publication Year',
    'Post-Publication'
]

assert validation_sample['month'].tolist() == months_to_check
assert validation_sample['period'].tolist() == expected_periods

print("Period validation checks passed!")
```

```python
# Compare Momentum returns across the three research periods

momentum = poc_data[
    (poc_data['signalname'] == 'Mom12m') &
    (poc_data['period'].isin([
        'Original Sample',
        'Pre-Publication Gap',
        'Post-Publication'
    ]))
]

# Calculate the average monthly return in each period
mom_summary = momentum.groupby('period')['ret'].agg(
    observations='count',
    avg_monthly_return='mean'
).reindex([
    'Original Sample',
    'Pre-Publication Gap',
    'Post-Publication'
])

display(mom_summary)

# Calculate the decline from the original sample to post-publication
original_return = mom_summary.loc['Original Sample', 'avg_monthly_return']
post_return = mom_summary.loc['Post-Publication', 'avg_monthly_return']

decline = (original_return - post_return) / original_return * 100

print(f"Momentum's profitability declined by {decline:.2f}% relative to its original sample.")
```

```python
# Save proof-of-concept dataset to outputs folder

from pathlib import Path

# Create outputs folder if it doesn't exist
output_dir = Path('outputs')
output_dir.mkdir(exist_ok=True)

# Save dataset
output_file = output_dir / 'checkpoint1_sample.csv'
poc_data.to_csv(output_file, index=False)

print(f"Dataset successfully saved to: {output_file}")
print(f"Observations: {len(poc_data):,}")
```
